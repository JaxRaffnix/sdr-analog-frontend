```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.fft as fft

plt.rcParams['figure.constrained_layout.use'] = True

sns.set_theme()
```

```python
fE = 500 
fM = 2000

T = 0.05
fs = 20*fM

t = np.arange(0, T, 1/fs)

xE = np.cos(2 * np.pi * fE * t)
xM = np.cos(2 * np.pi * fM * t)
xA = xE * xM
```

```python
fig, axes = plt.subplots(3, 1, sharex=True)

axes[0].plot(t, xE)
axes[0].set_title("Eingangssignal")

axes[1].plot(t, xM)
axes[1].set_title("Carrier")

axes[2].plot(t, xA)
axes[2].set_title("Ausgangssignal")

axes[-1].set_xlabel("Zeit (s)")

plt.show()
```

```python
yA = fft.fft(xA)
freqs = fft.fftfreq(len(xA), d=1/fs)

fig, ax = plt.subplots()
plt.plot(freqs, np.abs(yA))
ax.set_xlim(0, 3000)
ax.set(xlabel="Frequenz (Hz)", title="Frequenzspektrum")
plt.show()
```

```python
max_freq = freqs[np.argmax(np.abs(yA))]
print(f"Maximale Frequenz: {max_freq:.2f} Hz")
```
